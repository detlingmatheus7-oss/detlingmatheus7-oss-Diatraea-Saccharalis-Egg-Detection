```python
import os
import cv2
import shutil
import random
import albumentations as A

input_img_folder = '/content/drive/MyDrive/project/images/all'
input_label_folder = '/content/drive/MyDrive/project/labels/all'

output_img_folder = '/content/drive/MyDrive/projectA/images/all'
output_label_folder = '/content/drive/MyDrive/projectA/labels/all'

os.makedirs(output_img_folder, exist_ok=True)
os.makedirs(output_label_folder, exist_ok=True)

# Apenas transformações que não mexem na geometria
transform = A.Compose([
    A.RandomBrightnessContrast(p=0.4),
    A.HueSaturationValue(p=0.25),
    A.GaussNoise(p=0.00),
    A.GaussianBlur(p=0.00),
])

for filename in os.listdir(input_img_folder):

    if filename.endswith(".png"):

        # Apenas 30% das imagens serão aumentadas
        if random.random() > 0.40:
            continue

        img_path = os.path.join(input_img_folder, filename)
        label_name = filename.replace(".png", ".txt")

        label_path = os.path.join(input_label_folder, label_name)

        image = cv2.imread(img_path)

        augmented = transform(image=image)
        aug_image = augmented["image"]

        new_img_name = filename.replace(".png", "_aug.png")
        new_label_name = label_name.replace(".txt", "_aug.txt")

        cv2.imwrite(os.path.join(output_img_folder, new_img_name), aug_image)

        if os.path.exists(label_path):
            shutil.copy(
                label_path,
                os.path.join(output_label_folder, new_label_name)
            )
        else:
            print(f"⚠️ Label não encontrada: {label_name}")

print("Augmentação + cópia de labels concluída 🚀")
```

```python

from ultralytics import YOLO, RTDETR
from pathlib import Path
import pandas as pd
import re
import torch

# ============================================================
# 1. CONFIGURAÇÕES
# ============================================================

image_folder = Path('/content/drive/MyDrive/project/images/test')

CONF = 0.5

# Contagens manuais: NÃO ALTERAR
contagens_manuais = [83, 75, 43, 31, 41, 67, 44, 59]

# ============================================================
# 2. CAMINHOS DOS 16 MODELOS
# ============================================================
# Substitua os caminhos pelos caminhos reais dos seus pesos.

modelos = {
    # YOLO - 8 modelos
    'YOLO_01': ('YOLO', '/content/drive/MyDrive/project/model/weights/best.pt'),
    'YOLO_02': ('YOLO', '/content/drive/MyDrive/project/Modelo1b/weights/best.pt'),
    'YOLO_03': ('YOLO', '/content/drive/MyDrive/project/Modelo1c/weights/best.pt'),
    'YOLO_04': ('YOLO', '/content/drive/MyDrive/project/Modelo1d/weights/best.pt'),
    'YOLO_05': ('YOLO', '/content/drive/MyDrive/projectB/Modelo2a/weights/best.pt'),
    'YOLO_06': ('YOLO', '/content/drive/MyDrive/projectB/Modelo2b/weights/best.pt'),
    'YOLO_07': ('YOLO', '/content/drive/MyDrive/projectB/Modelo2c/weights/best.pt'),
    'YOLO_08': ('YOLO', '/content/drive/MyDrive/projectB/Modelo2d/weights/best.pt'),

    # RT-DETR - 8 modelos
    'RTDETR_01': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model1A/weights/best.pt'),
    'RTDETR_02': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model1B/weights/best.pt'),
    'RTDETR_03': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model1C/weights/best.pt'),
    'RTDETR_04': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model1D/weights/best.pt'),
    'RTDETR_05': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model2A/weights/best.pt'),
    'RTDETR_06': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model2B/weights/best.pt'),
    'RTDETR_07': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model2C/weights/best.pt'),
    'RTDETR_08': ('RTDETR', '/content/drive/MyDrive/RT-DETR/Model2D/weights/best.pt'),
}

# ============================================================
# 3. IDENTIFICAR AS IMAGENS DE TESTE
# ============================================================

def natural_key(path):
    return [
        int(part) if part.isdigit() else part.lower()
        for part in re.split(r'(\d+)', path.name)
    ]

extensoes = ['.png', '.jpg', '.jpeg', '.bmp']

image_paths = sorted(
    [
        p for p in image_folder.iterdir()
        if p.is_file()
        and p.suffix.lower() in extensoes
        and '_aug' not in p.stem.lower()
    ],
    key=natural_key
)

if len(image_paths) != len(contagens_manuais):
    raise ValueError(
        f'Foram encontradas {len(image_paths)} imagens, '
        f'mas existem {len(contagens_manuais)} contagens manuais.'
    )

print('Ordem das imagens utilizadas:')
for i, path in enumerate(image_paths):
    print(f'Imagem {i+1}: {path.name} | Manual: {contagens_manuais[i]}')

# ============================================================
# 4. TESTAR TODOS OS MODELOS
# ============================================================

resultados_modelos = {}
resumo_modelos = []

for nome, (tipo, caminho) in modelos.items():

    print('\n' + '=' * 60)
    print(f' TESTANDO: {nome}')
    print('=' * 60)

    # Carregar arquitetura correta
    if tipo == 'YOLO':
        model = YOLO(caminho)
    elif tipo == 'RTDETR':
        model = RTDETR(caminho)
    else:
        raise ValueError(f'Tipo de modelo desconhecido: {tipo}')

    # Fazer predições
    results = model.predict(
        source=[str(p) for p in image_paths],
        conf=CONF,
        verbose=False
    )

    dados = []

    for i, result in enumerate(results):

        # Número de caixas detectadas na imagem
        estimada = len(result.boxes) if result.boxes is not None else 0

        manual = contagens_manuais[i]

        erro_absoluto = abs(manual - estimada)

        erro_relativo = (erro_absoluto / manual) * 100

        dados.append({
            'ID da Imagem': i + 1,
            'Imagem': image_paths[i].name,
            'Contagem Manual': manual,
            'Contagem Estimada': estimada,
            'Erro Absoluto': erro_absoluto,
            'Erro Relativo (%)': round(erro_relativo, 2)
        })

    # Criar tabela do modelo
    df = pd.DataFrame(dados)

    erro_medio = df['Erro Relativo (%)'].mean()

    resultados_modelos[nome] = df

    resumo_modelos.append({
        'Modelo': nome,
        'Erro Relativo Médio (%)': round(erro_medio, 2)
    })

    print(df.to_string(index=False))
    print(f'\nErro Relativo Médio: {erro_medio:.2f}%')

    # Liberar memória
    del model
    del results

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# ============================================================
# 5. RESUMO COMPARATIVO DOS 16 MODELOS
# ============================================================

df_resumo = pd.DataFrame(resumo_modelos)

df_resumo = df_resumo.sort_values(
    by='Erro Relativo Médio (%)',
    ascending=True
)

print('\n' + '=' * 60)
print(' RESUMO COMPARATIVO DOS MODELOS')
print('=' * 60)

print(df_resumo.to_string(index=False))

# ============================================================
# 6. EXPORTAR TUDO PARA EXCEL
# ============================================================

arquivo_saida = '/content/drive/MyDrive/Erro_Relativo_16_Modelos.xlsx'

with pd.ExcelWriter(arquivo_saida, engine='openpyxl') as writer:

    # Uma aba para cada modelo
    for nome, df in resultados_modelos.items():
        df.to_excel(
            writer,
            sheet_name=nome,
            index=False
        )

        # Adicionar erro relativo médio abaixo da tabela
        ws = writer.sheets[nome]
        linha = len(df) + 3

        ws.cell(
            row=linha,
            column=1,
            value='Erro Relativo Médio (%)'
        )

        ws.cell(
            row=linha,
            column=2,
            value=round(df['Erro Relativo (%)'].mean(), 2)
        )

    # Aba de comparação geral
    df_resumo.to_excel(
        writer,
        sheet_name='Resumo Geral',
        index=False
    )

print('\nArquivo Excel salvo em:')
print(arquivo_saida)
```
